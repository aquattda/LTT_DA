## **1. Ôn tập lý thuyết**

**1) Quy trình khai phá dữ liệu CRISP – DM (Cross Industry Standard Process for Data Mining) là gì ? Quy trình khai phá dữ liệu SEMMA (Sample, Explore, Modify, Model, Access) là gì?**
- **CRISP-DM** là một bản kế hoạch chuẩn để thực hiện một dự án khoa học dữ liệu. Nó giống như một bản đồ chi tiết, chỉ các bước cần làm từ đầu đến cuối.
- Nó bao gồm 6 giai đoạn lặp đi lặp lại:

    - Hiểu bài toán kinh doanh (Business Understanding): Mục tiêu cuối cùng là gì? Cần giải quyết vấn đề gì?

    - Hiểu dữ liệu (Data Understanding): Thu thập và khám phá dữ liệu ban đầu. Dữ liệu có gì, chất lượng ra sao?

    - Chuẩn bị dữ liệu (Data Preparation): Đây là bước tốn thời gian nhất. Bao gồm làm sạch, tích hợp, và định dạng lại dữ liệu.

    - Mô hình hóa (Modeling): Lựa chọn và xây dựng các mô hình học máy (như Cây quyết định).

    - Đánh giá (Evaluation): Kiểm tra xem mô hình có đáp ứng được mục tiêu kinh doanh ở bước 1 không.

    - Triển khai (Deployment): Đưa mô hình vào sử dụng thực tế.

- **SEMMA** là một quy trình khai phá dữ liệu khác, thường được liên kết với phần mềm của SAS. Nó tập trung nhiều hơn vào các bước kỹ thuật của việc xây dựng mô hình. Tên của nó là viết tắt của 5 giai đoạn:

    - Sample (Lấy mẫu): Chọn một tập dữ liệu con đủ lớn nhưng vẫn có thể xử lý được.

    - Explore (Khám phá): Trực quan hóa, tìm kiếm các mối quan hệ và sự bất thường trong dữ liệu.

    - Modify (Chỉnh sửa): Chuẩn bị dữ liệu cho việc mô hình hóa (tạo biến mới, xử lý giá trị thiếu).

    - Model (Mô hình hóa): Áp dụng các thuật toán để tìm ra các mẫu.

    - Assess (Đánh giá): Đánh giá độ hiệu quả và hữu ích của mô hình.

>  **Khác biệt chính:** CRISP-DM bao quát toàn bộ dự án (cả khía cạnh kinh doanh), trong khi SEMMA tập trung chủ yếu vào các bước kỹ thuật để xây dựng mô hình.

**2) Cây quyết định hoạt động như thế nào? Hãy giải thích các thành phần chính (nút gốc, nút lá, nhánh)
và cách cây đưa ra dự đoán.**
- Decision Tree thực hiện chia dữ liệu thành các nhóm nhỏ, các nhóm nhỏ này sẽ dựa trên câu hỏi mà dự đoán câu trả lời. 

    - Nút gốc (Root Node): Là nơi chứa toàn bộ dữ liệu và nơi chứa câu hỏi quan trọng nhất. (ví dụ: "Đỗ đại học không?").

    - Nhánh (Branch): Là các câu trả lời cho câu hỏi (ví dụ: "Có" hoặc "Không"). Mỗi nhánh dẫn đến một nút khác.

    - Nút lá (Leaf Node): Đây là điểm kết thúc. Không trả lời câu hỏi mà là nơi đưa ra dự đoán cuối (ví dụ: "Đỗ đại học" hoặc "Không đỗ đại học").

> **Cách đưa ra dự đoán:** Đưa dự liều từ nút gốc, trả lời các câu hỏi cần thiết, dựa vào đó đi theo luồng của các nhánh và lặp đi lặp lại đến khi đến nút lá để đưa ra dự đoán.

**3) Các tiêu chí phân tách (splitting criteria) như Gini Index, Entropy, hay Information Gain được sử dụng trong cây quyết định là gì? Chúng khác nhau ra sao?**
- Tiêu chí phân tách là các công thức toán học giúp cây quyết định chọn câu hỏi tốt nhất để hỏi tại mỗi nút. "Tốt nhất" có nghĩa là câu hỏi đó chia dữ liệu thành các nhóm "thuần khiết" nhất. (Thuần khiết = các thành viên trong nhóm càng giống nhau càng tốt).

- Gini Index: Đo lường xác suất một phần tử được chọn ngẫu nhiên bị phân loại sai. Gini càng thấp càng tốt (độ "tạp nham" thấp).

- Entropy & Information Gain (Lợi ích thông tin):

    - Entropy: Đo lường mức độ "hỗn loạn" hoặc "ngẫu nhiên" của một nhóm. Entropy càng thấp, nhóm càng thuần khiết.

    - Information Gain: Đo lường mức giảm Entropy sau khi phân tách. Cây quyết định sẽ chọn cách phân tách có Information Gain cao nhất (tức là giảm được nhiều sự hỗn loạn nhất).

> **Khác biệt:** Cả hai đều dùng để đo độ "tinh khiết". Gini thường tính toán nhanh hơn một chút. Entropy có xu hướng tạo ra các cây cân bằng hơn.

**4) Rừng cây (Random Forest) là gì? Nó khác gì so với một cây quyết định đơn lẻ? Tại sao Random Forest thường có hiệu suất tốt hơn cây quyết định trong các bài toán phân loại?** 
- Random Forest (RF) là một mô hình học máy theo kiểu "quần thể" (ensemble). Thay vì chỉ xây dựng một cây quyết định, nó xây dựng hàng trăm (hoặc hàng nghìn) cây quyết định khác nhau rồi lấy ý kiến của số đông để đưa ra dự đoán cuối cùng.

- Khác biệt so với 1 cây quyết định (DT):

    - Số lượng: RF có nhiều cây, DT chỉ có một.

    - Cách học:

        Mỗi cây trong RF chỉ được học trên một tập dữ liệu con ngẫu nhiên (gọi là bootstrapping).

        Tại mỗi nút, mỗi cây chỉ được xem xét một tập đặc trưng con ngẫu nhiên để tìm cách phân tách tốt nhất.

> **Tại sao RF thường tốt hơn:** Một cây quyết định đơn lẻ rất dễ bị overfitting (học vẹt) – tức là nó học quá kỹ dữ liệu huấn luyện và dự đoán rất tệ trên dữ liệu mới.Random Forest tốt hơn vì nó dựa trên "trí tuệ tập thể". Bằng cách kết hợp dự đoán từ nhiều cây "đa dạng" (do tính ngẫu nhiên khi xây dựng), nó giảm thiểu đáng kể overfitting và đưa ra dự đoán tổng quát, ổn định hơn.

**5) Những ưu điểm và hạn chế của cây quyết định và Random Forest là gì? Trong trường hợp nào thì cây quyết định có thể hoạt động kém hiệu quả?**
**Cây quyết định (Decision Tree)**
- Ưu điểm:

    - Dễ hiểu và giải thích: Bạn có thể nhìn vào sơ đồ cây để hiểu tại sao nó đưa ra quyết định (mô hình "hộp trắng").

    - Không yêu cầu chuẩn hóa dữ liệu (scaling).

    - Xử lý được cả dữ liệu số và dữ liệu phân loại.

- Hạn chế:

    - Rất dễ bị Overfitting.

    - Không ổn định (một thay đổi nhỏ trong dữ liệu có thể tạo ra một cây hoàn toàn khác).

**Rừng cây (Random Forest)**

- Ưu điểm:

    - Hiệu suất rất cao và là một trong những mô hình tốt nhất cho dữ liệu dạng bảng.

    -Chống overfitting cực kỳ tốt.

    - Ổn định và mạnh mẽ.

- Hạn chế:

    - Mô hình "hộp đen" (Black box): Khó giải thích chính xác tại sao nó đưa ra một dự đoán cụ thể.

    - Tốn nhiều tài nguyên tính toán (thời gian và bộ nhớ) hơn để huấn luyện so với 1 cây.

> **Trường hợp DT kém hiệu quả:** Khi dữ liệu có nhiều nhiễu hoặc mối quan hệ giữa các biến phức tạp, DT sẽ cố gắng "học thuộc" cả nhiễu, dẫn đến overfitting và dự đoán sai trên thực tế.

**6) Viết đoạn code mẫu bằng Python (sử dụng Scikit-learn) để xây dựng một mô hình cây quyết địnhnkhông? Hãy mô tả các bước thực hiện**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Bước 1: Giả sử bạn có dữ liệu (X) và nhãn (y)
# Ví dụ: dự đoán việc đi chơi tennis (1) hay không (0)
# X (đặc trưng): [thời tiết, độ ẩm], y (nhãn): [đi chơi?]
X = pd.DataFrame({'thoi_tiet': [1, 0, 1, 0], 'do_am': [1, 1, 0, 0]}) # 1=nắng, 0=mưa; 1=cao, 0=thấp
y = pd.Series([1, 0, 1, 0])

# Bước 2: Chia dữ liệu thành tập huấn luyện và tập kiểm tra
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# Bước 3: Khởi tạo mô hình
# random_state để đảm bảo kết quả có thể tái lập
model_dt = DecisionTreeClassifier(random_state=42)

# Bước 4: Huấn luyện mô hình
model_dt.fit(X_train, y_train)

# Bước 5: Đưa ra dự đoán trên tập kiểm tra
y_pred = model_dt.predict(X_test)

# Bước 6: Đánh giá mô hình
accuracy = accuracy_score(y_test, y_pred)
print(f"Độ chính xác của Cây quyết định: {accuracy}")

# Dự đoán một mẫu mới: [thời tiết=nắng(1), độ ẩm=thấp(0)]
new_data = [[1, 0]]
print(f"Dự đoán cho mẫu mới: {model_dt.predict(new_data)}")
```

**7) Làm thế nào để triển khai một mô hình Random Forest trong Python? Bạn thường thiết lập các tham số nào (ví dụ: n_estimators, max_depth)?**





```python
from sklearn.ensemble import RandomForestClassifier

# Khởi tạo mô hình Random Forest
# n_estimators: số lượng cây trong rừng
# max_depth: độ sâu tối đa của mỗi cây
model_rf = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42)

# Các bước fit, predict, evaluate tương tự như Cây quyết định
model_rf.fit(X_train, y_train)
y_pred_rf = model_rf.predict(X_test)
print(f"Độ chính xác của Random Forest: {accuracy_score(y_test, y_pred_rf)}")
```

**8) Làm thế nào để đánh giá tầm quan trọng của các đặc trưng (feature importance) trong Random Forest bằng Python?** 
-  Random Forest có thể cho bạn biết đặc trưng nào (cột dữ liệu nào) quan trọng nhất trong việc đưa ra dự đoán. Sau khi huấn luyện (fit) mô hình, bạn chỉ cần truy cập thuộc tính .feature_importances_.



```python
# Giả sử model_rf đã được huấn luyện (fit)
importances = model_rf.feature_importances_

# Lấy tên các đặc trưng
feature_names = X.columns

# Tạo DataFrame để xem cho dễ
feature_importance_df = pd.DataFrame({'Feature': feature_names, 'Importance': importances})

# Sắp xếp từ cao đến thấp
feature_importance_df = feature_importance_df.sort_values(by='Importance', ascending=False)

print("\nTầm quan trọng của đặc trưng:")
print(feature_importance_df)
```

##### **2. Tạo cây quyết định và xác định các giá trị tham số có thể chọn để tìm tham số tối ưu**

```python
from sklearn.model_selection import GridSearchCV
params = {'max_depth':[1, 2, 4, 6, 8, 10, 12]} # các siêu tham số mat_depth (tránh overfitting)
dt = tree.DecisionTreeClassifier() # gọi model để grid gán các siêu tham số để huấn luyện
cv = GridSearchCV(dt, param_grid=params, scoring='roc_auc', # dùng AUC_ROC để đánh giá
n_jobs=None, refit=True, cv=4, verbose=1, # n_jobs = None tức là sử dụng toàn bộ CPU để chạy, dùng refit để khi chọn tham số tốt nhất để huấn luyện lại mô hình
error_score=np.nan,
return_train_score=True) # cv is the best model.
cv.fit(X_train, y_train)    
```

**Nhận xét:**

    - Ưu điểm: không phụ thuộc vào ngưỡng, ổn cho việc sắp xếp hạng rủi ro (đối với xác định ngưỡng cố đinh không khiến nghị)
- cv = 4: thực hiện chia dữ liệu train thành 4 phần (fold), mỗi phần sẽ lấy 1 fold làm validation còn 3 folds dùng fit.
- return_train_score = true: lưu điểm trên tập train của mỗi fold, điều này làm tốn thời gian nên default = False. Tuy nhiên nó giúp chuẩn đoán bias và variance

##### **3. Vẽ biểu đồ đánh giá mô hình với các tham số chiều sâu của cây khác nhau**

```python
cv_results_df = pd.DataFrame(cv.cv_results_)
#View the names of the remaining columns in the results DataFrame
cv_results_df.columns
ax = plt.axes()
ax.errorbar(cv_results_df['param_max_depth'],
cv_results_df['mean_train_score'],
yerr=cv_results_df['std_train_score']/np.sqrt(4),
label='Mean $m$ 1 SE training scores')
ax.errorbar(cv_results_df['param_max_depth'],
cv_results_df['mean_test_score'],
yerr=cv_results_df['std_test_score']/np.sqrt(4),
label='Mean $pm$ 1 SE testing scores')
ax.legend()
plt.xlabel('max_depth')
plt.ylabel('ROC AUC')
ax.legend()
plt.xlabel('max_depth')
plt.ylabel('ROC AUC')
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
import numpy as np

rf = RandomForestClassifier(
    random_state=42,
    n_jobs=-1,           # cho RF (không phải GridSearch)
    class_weight=None    # hoặc 'balanced' nếu lệch lớp
)

rf_params_ex = {'n_estimators': list(range(10, 110, 10))}

cv_rf_ex = GridSearchCV(
    rf,
    param_grid=rf_params_ex,
    scoring='roc_auc',   # nhị phân; đa lớp dùng 'roc_auc_ovr'
    n_jobs=-1,           # song song hóa grid search
    refit=True,
    cv=4,
    verbose=1,
    error_score=np.nan,  # <— KHÔNG dùng 'np.nan'
    return_train_score=True
)

cv_rf_ex.fit(X_train, y_train)
```

```python
cv_results_df = pd.DataFrame(cv.cv_results_)
#View the names of the remaining columns in the results DataFrame
cv_results_df.columns
ax = plt.axes()
ax.errorbar(cv_results_df['param_max_depth'],
cv_results_df['mean_train_score'],
yerr=cv_results_df['std_train_score']/np.sqrt(4),
label='Mean $m$ 1 SE training scores')
ax.errorbar(cv_results_df['param_max_depth'],
cv_results_df['mean_test_score'],
yerr=cv_results_df['std_test_score']/np.sqrt(4),
label='Mean $pm$ 1 SE testing scores')
ax.legend()
plt.xlabel('max_depth')
plt.ylabel('ROC AUC')
ax.legend()
plt.xlabel('max_depth')
plt.ylabel('ROC AUC')
plt.show()
```

- So sánh kết quả đánh giá mô hình với chiều sâu khác nhau. Với độ sâu quá cao khi model học tham số sẽ có ROC AUC cao chứng tỏ độ phức tạp của model cao. Ngược lại thì khi test kết quả model học thì do độ phức tạp của dữ liệu thấp hơn khiến kết quả bias.
- Điều này muốn chứng minh rằng không phải độ sâu càng cao thì kết quả trả về sẽ cao, có khi nó lại làm ngược lại

#### **6. Tạo rừng cây với lớp RandomForestClassifier**

##### 6.1 Xây dựng model 

```python
# import thư viện
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import accuracy_score, roc_auc_score
# pipline
rf_pipe = Pipeline(steps=[
    ("imp", SimpleImputer(strategy="median")), # điền các giá trị của cột bằng median (use train  to test), đảm bảo đúng đắn.
    ("clf", RandomForestClassifier(
        n_estimators=300, # số cây (nhiều -> tốn  time)
        max_depth=None, # độ sâu của cây không giới hạn (sâu đến khi node lá đủ)
        min_samples_split=2, # số mẫu tối thiểu của node được tách (tăng -> giảm overfit)
        min_samples_leaf=1, # số mẫu tối thiểu ở mỗi lá (giảm overfit)
        max_features="sqrt", # tại mỗi lần tách chỉ xét sqrt of features (tăng tính đa dạng giữa các cây -> classification)
        bootstrap=True, # mỗi cây được huấn luyện trên booststrap giúp giảm phương sai
        random_state=42, 
        n_jobs=-1 
    ))
])

rf_pipe.fit(X_train, y_train)
y_pred = rf_pipe.predict(X_test)
y_proba = rf_pipe.predict_proba(X_test)[:, 1] # xác suất mô hình ước lượng cho lable = 1 trên từng rows test

print("Accuracy:", accuracy_score(y_test, y_pred))
print("ROC AUC :", roc_auc_score(y_test, y_proba))
```

**Nhận xét:**
- AUC (0.731) sẽ tính bằng cách quét mọi ngưỡng có trong y_proba (1->0) để vẽ ROC. Vậy có ~73.1% các cặp âm dương được xếp đúng thứ tự cho nên mô hình không phụ thuộc bởi ngưỡng.
- Accuracy (65%) đây là tỷ lệ chính xác mô hình đã dược huấn luyện. Cho thấy sau khi huấn luyện mô hình có thể dự đoán khá ổn.

##### 6.2 Tối ưu số cây bằng GridSearch

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "clf__n_estimators": [50, 100, 200, 300, 500] # siêu tham số
}

cv_rf = GridSearchCV(
    estimator=rf_pipe,
    param_grid=param_grid,
    scoring="roc_auc",
    cv=4, # chia 4 fold cho mỗi tập train (láy 3 fit 1 làm validation)
    n_jobs=-1,
    verbose=1,
    return_train_score=True
)
cv_rf.fit(X_train, y_train)

print("Best params :", cv_rf.best_params_)
print("Best CV AUC :", cv_rf.best_score_)

# Dùng model tốt nhất để đánh giá trên tập validation
best_rf = cv_rf.best_estimator_
y_pred = best_rf.predict(X_test)
y_proba = best_rf.predict_proba(X_test)[:, 1]
print("Val Accuracy:", accuracy_score(y_test, y_pred))
print("Val ROC AUC :", roc_auc_score(y_test, y_proba))
```

**Nhận xét:**
- Sau khi tìm tham số tối ưu thì có thể thấy số cây phù hợp là 100 cây
- Khi đó chỉ số AUC phù hợp nhất là 72%, do chọn tối ưu cho nên có thể giảm % đôi chút tuy nhiên dữ liệu vẫn không phụ thuộc ngưỡng.

#### **5. Tạo rừng cây với lớp RandomForestClassifier**

##### **5.1 Xây dựng model** 

```python
# import thư viện
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import accuracy_score, roc_auc_score
# pipline
rf_pipe = Pipeline(steps=[
    ("imp", SimpleImputer(strategy="median")), # điền các giá trị của cột bằng median (use train  to test), đảm bảo đúng đắn.
    ("clf", RandomForestClassifier(
        n_estimators=300, # số cây (nhiều -> tốn  time)
        max_depth=None, # độ sâu của cây không giới hạn (sâu đến khi node lá đủ)
        min_samples_split=2, # số mẫu tối thiểu của node được tách (tăng -> giảm overfit)
        min_samples_leaf=1, # số mẫu tối thiểu ở mỗi lá (giảm overfit)
        max_features="sqrt", # tại mỗi lần tách chỉ xét sqrt of features (tăng tính đa dạng giữa các cây -> classification)
        bootstrap=True, # mỗi cây được huấn luyện trên booststrap giúp giảm phương sai
        random_state=42, 
        n_jobs=-1 
    ))
])

rf_pipe.fit(X_train, y_train)
y_pred = rf_pipe.predict(X_test)
y_proba = rf_pipe.predict_proba(X_test)[:, 1] # xác suất mô hình ước lượng cho lable = 1 trên từng rows test

print("Accuracy:", accuracy_score(y_test, y_pred))
print("ROC AUC :", roc_auc_score(y_test, y_proba))
```

**Nhận xét:**
- AUC (0.80) sẽ tính bằng cách quét mọi ngưỡng có trong y_proba (1->0) để vẽ ROC. Vậy có ~80% các cặp âm dương được xếp đúng thứ tự cho nên mô hình không phụ thuộc bởi ngưỡng.
- Accuracy (75%) đây là tỷ lệ chính xác mô hình đã dược huấn luyện. Cho thấy sau khi huấn luyện mô hình có thể dự đoán khá ổn.

##### **5.2 Tối ưu số cây bằng GridSearch**

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "clf__n_estimators": [50, 100, 200, 300, 500] # siêu tham số
}

cv_rf = GridSearchCV(
    estimator=rf_pipe,
    param_grid=param_grid,
    scoring="roc_auc",
    cv=4, # chia 4 fold cho mỗi tập train (láy 3 fit 1 làm validation)
    n_jobs=-1,
    verbose=1,
    return_train_score=True
)
cv_rf.fit(X_train, y_train)

print("Best params :", cv_rf.best_params_)
print("Best CV AUC :", cv_rf.best_score_)

# Dùng model tốt nhất để đánh giá trên tập validation
best_rf = cv_rf.best_estimator_
y_pred = best_rf.predict(X_test)
y_proba = best_rf.predict_proba(X_test)[:, 1]
print("Val Accuracy:", accuracy_score(y_test, y_pred))
print("Val ROC AUC :", roc_auc_score(y_test, y_proba))
```

**Nhận xét:**
- Sau khi tìm tham số tối ưu thì có thể thấy số cây phù hợp là 500 cây
- Khi đó chỉ số AUC phù hợp nhất là 84%, do chọn tối ưu cho nên có thể giảm hoặc tăng % đôi chút tuy nhiên dữ liệu vẫn không phụ thuộc ngưỡng.
