```python
# Hồi quy Power-law: P(k) ~ k^(-gamma)
# Sử dụng log transformation: log(P(k)) = -gamma * log(k) + c

# Lọc bỏ các giá trị 0
degree_values_nonzero = [d for d in degree_values if d > 0 and degree_counts[d] > 0]
degree_freq_nonzero = [degree_counts[d] for d in degree_values_nonzero]

if len(degree_values_nonzero) > 1:
    # Chuyển sang log scale
    log_degrees = np.log10(degree_values_nonzero)
    log_freq = np.log10(degree_freq_nonzero)
    
    # Hồi quy tuyến tính
    slope, intercept, r_value, p_value, std_err = stats.linregress(log_degrees, log_freq)
    
    print("=" * 60)
    print("HỒI QUY POWER-LAW")
    print("=" * 60)
    print(f"\nMô hình: P(k) = C * k^γ")
    print(f"\n- Hệ số gamma (γ): {-slope:.4f}")
    print(f"- Hệ số tương quan R²: {r_value**2:.4f}")
    print(f"- P-value: {p_value:.6f}")
    
    # Vẽ đồ thị hồi quy
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.scatter(degree_values_nonzero, degree_freq_nonzero, alpha=0.6, s=50, label='Dữ liệu thực')
    
    # Đường hồi quy
    x_fit = np.logspace(np.log10(min(degree_values_nonzero)), 
                        np.log10(max(degree_values_nonzero)), 100)
    y_fit = 10**(slope * np.log10(x_fit) + intercept)
    ax.plot(x_fit, y_fit, 'r--', linewidth=2, 
            label=f'Power-law fit: γ={-slope:.2f}, R²={r_value**2:.3f}')
    
    ax.set_xlabel('Bậc (log scale)', fontsize=12)
    ax.set_ylabel('Tần suất (log scale)', fontsize=12)
    ax.set_title('Hồi quy Power-law cho Phân phối Bậc', fontsize=14, fontweight='bold')
    ax.set_xscale('log')
    ax.set_yscale('log')
    ax.legend()
    ax.grid(True, alpha=0.3, which='both')
    plt.tight_layout()
    plt.show()
    
    print(f"\nNhận xét:")
    if r_value**2 > 0.8:
        print(f"- Mạng tuân theo phân phối power-law rất tốt (R² = {r_value**2:.3f})")
        print(f"- Đây là mạng scale-free với gamma = {-slope:.2f}")
    elif r_value**2 > 0.6:
        print(f"- Mạng có xu hướng power-law khá tốt (R² = {r_value**2:.3f})")
    else:
        print(f"- Mạng không tuân theo phân phối power-law rõ ràng (R² = {r_value**2:.3f})")
    
    print("\n" + "=" * 60)
else:
    print("Không đủ dữ liệu để thực hiện hồi quy power-law")
```

## 3.5 Phân tích tính tương đương/tương tự về cấu trúc nút

```python
# Structural Equivalence - Sử dụng Jaccard similarity
print("=" * 60)
print("PHÂN TÍCH TÍNH TƯƠNG ĐƯƠNG CẤU TRÚC")
print("=" * 60)

def jaccard_similarity(g, node1, node2):
    """Tính Jaccard similarity giữa hai nút dựa trên tập láng giềng"""
    neighbors1 = set(g.neighbors(node1))
    neighbors2 = set(g.neighbors(node2))
    
    if len(neighbors1) == 0 and len(neighbors2) == 0:
        return 0
    
    intersection = len(neighbors1.intersection(neighbors2))
    union = len(neighbors1.union(neighbors2))
    
    return intersection / union if union > 0 else 0

# Chọn một số nút để phân tích (ví dụ: top 20 nút có degree cao nhất)
sample_nodes = list(centrality_df.nlargest(20, 'Degree')['Node'])

print(f"\nTính toán Jaccard Similarity cho {len(sample_nodes)} nút có bậc cao nhất...")

# Tính ma trận similarity
similarity_matrix = np.zeros((len(sample_nodes), len(sample_nodes)))

for i, node1 in enumerate(sample_nodes):
    for j, node2 in enumerate(sample_nodes):
        if i <= j:
            sim = jaccard_similarity(G, node1, node2)
            similarity_matrix[i, j] = sim
            similarity_matrix[j, i] = sim

# Vẽ heatmap
fig, ax = plt.subplots(figsize=(12, 10))
im = ax.imshow(similarity_matrix, cmap='YlOrRd', aspect='auto')

# Cấu hình axes
ax.set_xticks(range(len(sample_nodes)))
ax.set_yticks(range(len(sample_nodes)))
ax.set_xticklabels(sample_nodes, rotation=45, ha='right')
ax.set_yticklabels(sample_nodes)

# Thêm colorbar
cbar = plt.colorbar(im, ax=ax)
cbar.set_label('Jaccard Similarity', rotation=270, labelpad=20)

ax.set_title('Ma trận Tương đồng Cấu trúc (Jaccard Similarity)\ngiữa các Nút có Bậc cao', 
             fontsize=14, fontweight='bold', pad=20)

plt.tight_layout()
plt.show()

# Tìm các cặp nút tương đồng nhất
print("\nCác cặp nút có cấu trúc tương đồng cao (top 10):")
print("-" * 60)
similar_pairs = []
for i in range(len(sample_nodes)):
    for j in range(i+1, len(sample_nodes)):
        if similarity_matrix[i, j] > 0:
            similar_pairs.append((sample_nodes[i], sample_nodes[j], similarity_matrix[i, j]))

similar_pairs.sort(key=lambda x: x[2], reverse=True)
for rank, (node1, node2, sim) in enumerate(similar_pairs[:10], 1):
    print(f"   {rank:2d}. Nút {node1} <-> Nút {node2}: {sim:.4f}")

print("\n" + "=" * 60)
```

## 4.2 Phát hiện cộng đồng - So sánh các thuật toán

```python
# Import thêm thư viện cho community detection
from networkx.algorithms import community

print("=" * 60)
print("PHÁT HIỆN CỘNG ĐỒNG - SO SÁNH CÁC THUẬT TOÁN")
print("=" * 60)

# Dictionary để lưu kết quả
community_results = {}

# 1. Greedy Modularity Communities
print("\n1. Chạy Greedy Modularity Maximization...")
greedy_communities = list(community.greedy_modularity_communities(G))
community_results['Greedy'] = greedy_communities

# 2. Label Propagation
print("2. Chạy Label Propagation...")
label_prop_communities = list(community.label_propagation_communities(G))
community_results['Label Propagation'] = label_prop_communities

# 3. Louvain (nếu có python-louvain)
try:
    import community as community_louvain
    print("3. Chạy Louvain...")
    louvain_partition = community_louvain.best_partition(G)
    # Convert to list of sets
    louvain_communities = {}
    for node, comm_id in louvain_partition.items():
        if comm_id not in louvain_communities:
            louvain_communities[comm_id] = set()
        louvain_communities[comm_id].add(node)
    louvain_communities = list(louvain_communities.values())
    community_results['Louvain'] = louvain_communities
except ImportError:
    print("3. Louvain không có sẵn (cần cài đặt: pip install python-louvain)")
    louvain_communities = None

# 4. Girvan-Newman (chỉ chạy nếu mạng không quá lớn)
if G.number_of_nodes() <= 200:
    print("4. Chạy Girvan-Newman (có thể mất thời gian)...")
    gn_communities_generator = community.girvan_newman(G)
    # Lấy phân vùng đầu tiên (hoặc có thể lấy nhiều level)
    gn_communities = next(gn_communities_generator)
    gn_communities = [set(c) for c in gn_communities]
    community_results['Girvan-Newman'] = gn_communities
else:
    print("4. Girvan-Newman bỏ qua (mạng quá lớn)")
    gn_communities = None

print("\n✓ Hoàn thành phát hiện cộng đồng!")
print("=" * 60)
```

### So sánh kết quả các thuật toán

```python
# So sánh số lượng cộng đồng và modularity
print("=" * 60)
print("SO SÁNH CÁC THUẬT TOÁN PHÁT HIỆN CỘNG ĐỒNG")
print("=" * 60)

comparison_data = []

for method_name, communities in community_results.items():
    num_communities = len(communities)
    
    # Tính kích thước cộng đồng
    sizes = [len(c) for c in communities]
    avg_size = np.mean(sizes)
    max_size = max(sizes)
    min_size = min(sizes)
    
    # Tính modularity
    modularity_score = community.modularity(G, communities)
    
    comparison_data.append({
        'Thuật toán': method_name,
        'Số cộng đồng': num_communities,
        'Kích thước TB': f'{avg_size:.1f}',
        'Kích thước Min': min_size,
        'Kích thước Max': max_size,
        'Modularity': f'{modularity_score:.4f}'
    })
    
    print(f"\n{method_name}:")
    print(f"  - Số cộng đồng: {num_communities}")
    print(f"  - Kích thước trung bình: {avg_size:.1f}")
    print(f"  - Kích thước min-max: {min_size} - {max_size}")
    print(f"  - Modularity: {modularity_score:.4f}")

# Tạo bảng so sánh
comparison_df = pd.DataFrame(comparison_data)
print("\n" + "=" * 60)
print("BẢNG SO SÁNH:")
print("=" * 60)
print(comparison_df.to_string(index=False))
print("\n" + "=" * 60)

# Visualization so sánh
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# So sánh số cộng đồng
methods = [d['Thuật toán'] for d in comparison_data]
num_comms = [d['Số cộng đồng'] for d in comparison_data]
modularities = [float(d['Modularity']) for d in comparison_data]

axes[0].bar(methods, num_comms, edgecolor='black', alpha=0.7)
axes[0].set_ylabel('Số cộng đồng', fontsize=12)
axes[0].set_title('Số cộng đồng phát hiện được', fontsize=14, fontweight='bold')
axes[0].tick_params(axis='x', rotation=45)
axes[0].grid(True, alpha=0.3, axis='y')

# So sánh Modularity
colors = ['green' if m == max(modularities) else 'steelblue' for m in modularities]
axes[1].bar(methods, modularities, color=colors, edgecolor='black', alpha=0.7)
axes[1].set_ylabel('Modularity', fontsize=12)
axes[1].set_title('Chất lượng (Modularity) - Cao hơn là tốt hơn', fontsize=14, fontweight='bold')
axes[1].tick_params(axis='x', rotation=45)
axes[1].grid(True, alpha=0.3, axis='y')
axes[1].axhline(y=0.3, color='red', linestyle='--', linewidth=1, alpha=0.5, label='Ngưỡng tốt (0.3)')
axes[1].legend()

plt.tight_layout()
plt.show()
```

### Phân tích chi tiết cộng đồng (chọn thuật toán tốt nhất)

```python
# Chọn thuật toán có modularity cao nhất để phân tích chi tiết
best_method = max(comparison_data, key=lambda x: float(x['Modularity']))['Thuật toán']
best_communities = community_results[best_method]

print("=" * 60)
print(f"PHÂN TÍCH CHI TIẾT - THUẬT TOÁN TỐT NHẤT: {best_method}")
print("=" * 60)

# Thống kê kích thước các cộng đồng
community_sizes = sorted([len(c) for c in best_communities], reverse=True)

print(f"\nTOP 10 cộng đồng lớn nhất:")
print("-" * 60)
for i, size in enumerate(community_sizes[:10], 1):
    percentage = (size / G.number_of_nodes()) * 100
    print(f"   {i:2d}. Cộng đồng: {size} nút ({percentage:.1f}%)")

# Vẽ phân phối kích thước cộng đồng
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Histogram
axes[0].hist(community_sizes, bins=min(30, len(community_sizes)), 
             edgecolor='black', alpha=0.7)
axes[0].axvline(np.mean(community_sizes), color='red', linestyle='--', 
                linewidth=2, label=f'Trung bình = {np.mean(community_sizes):.1f}')
axes[0].set_xlabel('Kích thước cộng đồng', fontsize=12)
axes[0].set_ylabel('Tần suất', fontsize=12)
axes[0].set_title(f'Phân phối Kích thước Cộng đồng ({best_method})', 
                  fontsize=14, fontweight='bold')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Top communities
top_n = min(15, len(community_sizes))
axes[1].barh(range(top_n), community_sizes[:top_n], edgecolor='black', alpha=0.7)
axes[1].set_xlabel('Số nút', fontsize=12)
axes[1].set_ylabel('Cộng đồng (xếp hạng)', fontsize=12)
axes[1].set_title(f'Top {top_n} Cộng đồng lớn nhất', fontsize=14, fontweight='bold')
axes[1].invert_yaxis()
axes[1].grid(True, alpha=0.3, axis='x')

plt.tight_layout()
plt.show()

print("\n" + "=" * 60)
```

## 4.3 Visualization cộng đồng trên bố cục mạng

```python
# Visualization cộng đồng
print("Đang tạo visualization cộng đồng...")

# Tạo màu cho mỗi cộng đồng
import matplotlib.cm as cm

# Tạo dict ánh xạ node -> community id
node_to_community = {}
for comm_id, comm in enumerate(best_communities):
    for node in comm:
        node_to_community[node] = comm_id

# Tạo màu sắc
num_communities = len(best_communities)
colors = cm.tab20(np.linspace(0, 1, num_communities))

# Nếu có quá nhiều cộng đồng, sử dụng colormap khác
if num_communities > 20:
    colors = cm.viridis(np.linspace(0, 1, num_communities))

# Tạo layout
pos = nx.spring_layout(G, k=0.5, iterations=50, seed=42)

# Vẽ đồ thị với màu theo cộng đồng
fig, axes = plt.subplots(1, 2, figsize=(20, 10))

# Subplot 1: Colored by community
node_colors = [colors[node_to_community[node]] for node in G.nodes()]
node_sizes = [G.degree(n) * 20 for n in G.nodes()]

nx.draw_networkx_nodes(G, pos, ax=axes[0], node_color=node_colors, 
                       node_size=node_sizes, alpha=0.8)
nx.draw_networkx_edges(G, pos, ax=axes[0], edge_color='gray', 
                       alpha=0.3, width=0.5)

axes[0].set_title(f'Cộng đồng ({best_method}) - {num_communities} cộng đồng\nMàu sắc thể hiện cộng đồng', 
                  fontsize=14, fontweight='bold')
axes[0].axis('off')

# Subplot 2: Chỉ hiển thị top 5 cộng đồng lớn nhất với label
top_5_communities = sorted(enumerate(best_communities), 
                          key=lambda x: len(x[1]), reverse=True)[:5]

for ax_idx, (comm_id, comm) in enumerate(top_5_communities):
    # Tạo subgraph cho cộng đồng này
    subgraph_nodes = list(comm)
    color = colors[comm_id]
    
    # Vẽ các nút của cộng đồng này
    node_list = [n for n in subgraph_nodes if n in G]
    if node_list:
        sizes = [G.degree(n) * 20 for n in node_list]
        nx.draw_networkx_nodes(G, pos, ax=axes[1], nodelist=node_list,
                              node_color=[color], node_size=sizes, 
                              alpha=0.8, label=f'C{ax_idx+1} ({len(comm)} nút)')

nx.draw_networkx_edges(G, pos, ax=axes[1], edge_color='gray', 
                       alpha=0.3, width=0.5)

axes[1].set_title(f'Top 5 Cộng đồng lớn nhất', fontsize=14, fontweight='bold')
axes[1].legend(loc='best', fontsize=10)
axes[1].axis('off')

plt.tight_layout()
plt.show()

print(f"✓ Visualization hoàn thành!")
```

### Xuất kết quả cộng đồng cho Gephi

```python
# Thêm thuộc tính community vào đồ thị để xuất ra Gephi
G_with_communities = G.copy()

# Thêm community attribute cho từng node
for node in G_with_communities.nodes():
    G_with_communities.nodes[node]['community'] = node_to_community[node]
    G_with_communities.nodes[node]['degree'] = G.degree(node)

# Xuất ra file
output_file_communities = 'network_with_communities.gexf'
nx.write_gexf(G_with_communities, output_file_communities)

print(f"✓ Đã xuất file: {output_file_communities}")
print("\nHướng dẫn sử dụng trong Gephi:")
print("1. Mở file trong Gephi")
print("2. Trong Appearance panel:")
print("   - Nodes > Partition > community (chọn màu cho mỗi cộng đồng)")
print("   - Nodes > Ranking > degree (chọn size theo bậc)")
print("3. Chạy Layout: ForceAtlas 2")
print("4. Chạy Modularity trong Statistics panel để xác nhận kết quả")
```

## 4.4 Phân tích mối quan hệ giữa các cộng đồng

```python
# Phân tích inter-community edges
print("=" * 60)
print("PHÂN TÍCH MỐI QUAN HỆ GIỮA CÁC CỘNG ĐỒNG")
print("=" * 60)

# Đếm số cạnh giữa các cộng đồng
inter_community_edges = 0
intra_community_edges = 0

for u, v in G.edges():
    if node_to_community[u] == node_to_community[v]:
        intra_community_edges += 1
    else:
        inter_community_edges += 1

total_edges = G.number_of_edges()

print(f"\n- Tổng số cạnh: {total_edges}")
print(f"- Cạnh trong cộng đồng: {intra_community_edges} ({intra_community_edges/total_edges*100:.1f}%)")
print(f"- Cạnh giữa các cộng đồng: {inter_community_edges} ({inter_community_edges/total_edges*100:.1f}%)")

# Tạo ma trận kết nối giữa các cộng đồng (cho top communities)
top_n_comms = min(10, len(best_communities))
top_comm_ids = [i for i, _ in sorted(enumerate(best_communities), 
                                     key=lambda x: len(x[1]), reverse=True)[:top_n_comms]]

# Ma trận kết nối
connection_matrix = np.zeros((top_n_comms, top_n_comms))

for u, v in G.edges():
    comm_u = node_to_community[u]
    comm_v = node_to_community[v]
    
    if comm_u in top_comm_ids and comm_v in top_comm_ids:
        idx_u = top_comm_ids.index(comm_u)
        idx_v = top_comm_ids.index(comm_v)
        connection_matrix[idx_u, idx_v] += 1
        if idx_u != idx_v:  # Nếu không phải cùng cộng đồng
            connection_matrix[idx_v, idx_u] += 1

# Vẽ heatmap
fig, ax = plt.subplots(figsize=(10, 8))

im = ax.imshow(connection_matrix, cmap='YlOrRd', aspect='auto')

# Labels
community_labels = [f'C{i+1}\n({len(best_communities[top_comm_ids[i]])} nút)' 
                   for i in range(top_n_comms)]
ax.set_xticks(range(top_n_comms))
ax.set_yticks(range(top_n_comms))
ax.set_xticklabels(community_labels, rotation=45, ha='right')
ax.set_yticklabels(community_labels)

# Colorbar
cbar = plt.colorbar(im, ax=ax)
cbar.set_label('Số cạnh kết nối', rotation=270, labelpad=20)

# Thêm giá trị vào các ô
for i in range(top_n_comms):
    for j in range(top_n_comms):
        text = ax.text(j, i, int(connection_matrix[i, j]),
                      ha="center", va="center", color="black", fontsize=9)

ax.set_title(f'Ma trận Kết nối giữa Top {top_n_comms} Cộng đồng lớn nhất', 
             fontsize=14, fontweight='bold', pad=20)

plt.tight_layout()
plt.show()

print("\n" + "=" * 60)
```
