## 📊 Giới thiệu về mạng phân tích

### Loại mạng: **Mạng Scale-Free (Barabási-Albert Model)**

### Nguồn dữ liệu: **Mô hình sinh Barabási-Albert**

Mô hình **Barabási-Albert (BA)** là một mô hình toán học để sinh mạng ngẫu nhiên có cấu trúc scale-free, mô phỏng cách các mạng thực tế phát triển và phát triển qua thời gian.

### 🔬 Đặc điểm của mô hình BA:

1. **Cơ chế Preferential Attachment (Gắn kết ưu tiên)**:
   - Nút mới có xu hướng kết nối với các nút đã có bậc cao
   - "Người giàu ngày càng giàu" - The rich get richer
   - Mô phỏng hiện tượng: người nổi tiếng thu hút nhiều follower hơn

2. **Phân phối Power-Law**:
   - P(k) ~ k^(-γ) với γ ≈ 2.5 - 3.0
   - Có vài "hub nodes" với bậc rất cao
   - Phần lớn các nút có bậc thấp

3. **Tính chất Small-World**:
   - Đường đi trung bình ngắn (3-5 bước)
   - Hiện tượng "6 bậc phân cách" (Six Degrees of Separation)

### 📈 Tham số mạng trong bài:

```python
G = nx.barabasi_albert_graph(n=300, m=3, seed=42)
```

- **n = 300**: Tổng số nút (mô phỏng 300 người dùng)
- **m = 3**: Mỗi nút mới kết nối với 3 nút hiện có
- **seed = 42**: Seed để tái tạo kết quả

### 🎯 Mục tiêu phân tích:

1. **Phân tích cấu trúc tổng quan**:
   - Xác nhận tính chất scale-free (power-law distribution)
   - Đo lường các chỉ số: đường kính, hệ số phân cụm, độ dài đường đi

2. **Tìm các nút quan trọng**:
   - Xác định "hub nodes" (người có ảnh hưởng lớn)
   - Phân loại nút: Hub, Bridge, Peripheral
   - So sánh các độ đo centrality

3. **Phát hiện cộng đồng**:
   - Tìm các nhóm/cộng đồng trong mạng
   - So sánh các thuật toán phát hiện cộng đồng
   - Phân tích mối quan hệ giữa các cộng đồng

### 🌐 Ứng dụng thực tế:

Mô hình BA áp dụng cho:
- **Mạng xã hội**: Facebook, Twitter (follower networks)
- **Mạng Internet**: Kết nối giữa các router
- **Mạng trích dẫn**: Trích dẫn giữa các bài báo khoa học
- **Mạng sinh học**: Mạng protein-protein interactions

## 2.2 Kiểu đồ thị và Thuộc tính cơ bản

````python
# Xác định kiểu đồ thị
print("="*60)
print("THÔNG TIN CƠ BẢN VỀ MẠNG")
print("="*60)

# Kiểm tra loại đồ thị
is_directed = nx.is_directed(G)
is_weighted = nx.is_weighted(G)

print(f"\n1. KIỂU ĐỒ THỊ:")
print(f"   - Có hướng: {'Có' if is_directed else 'Không'}")
print(f"   - Có trọng số: {'Có' if is_weighted else 'Không'}")
print(f"   - Loại đồ thị: {type(G).__name__}")

# Số lượng nút và cạnh
num_nodes = G.number_of_nodes()
num_edges = G.number_of_edges()

print(f"\n2. KÍ CH THƯỚC:")
print(f"   - Số nút: {num_nodes}")
print(f"   - Số cạnh: {num_edges}")
print(f"   - Mật độ: {nx.density(G):.6f}")

# Kiểm tra tính liên thông
if is_directed:
    is_connected = nx.is_weakly_connected(G)
    print(f"\n3. TÍNH LIÊN THÔNG:")
    print(f"   - Liên thông yếu: {is_connected}")
    print(f"   - Số thành phần liên thông yếu: {nx.number_weakly_connected_components(G)}")
else:
    is_connected = nx.is_connected(G)
    print(f"\n3. TÍNH LIÊN THÔNG:")
    print(f"   - Liên thông: {is_connected}")
    print(f"   - Số thành phần liên thông: {nx.number_connected_components(G)}")
    
# Lấy thành phần liên thông lớn nhất (largest connected component)
if not is_connected:
    if is_directed:
        largest_cc = max(nx.weakly_connected_components(G), key=len)
    else:
        largest_cc = max(nx.connected_components(G), key=len)
    G_lcc = G.subgraph(largest_cc).copy()
    print(f"   - Kích thước thành phần lớn nhất: {G_lcc.number_of_nodes()} nút")
else:
    G_lcc = G
    print(f"   - Đồ thị đã liên thông hoàn toàn")

print("\n" + "="*60)
````

## 2.3 Đường kính và Bán kính

````python
# Tính đường kính và bán kính (chỉ cho đồ thị liên thông)
print("="*60)
print("ĐƯỜNG KÍNH VÀ BÁN KÍNH")
print("="*60)

if nx.is_connected(G_lcc):
    diameter = nx.diameter(G_lcc)
    radius = nx.radius(G_lcc)
    
    print(f"\n- Đường kính (Diameter): {diameter}")
    print(f"- Bán kính (Radius): {radius}")
    
    # Tìm center và periphery
    center = nx.center(G_lcc)
    periphery = nx.periphery(G_lcc)
    
    print(f"\n- Số nút ở trung tâm (center): {len(center)}")
    print(f"- Số nút ở ngoại vi (periphery): {len(periphery)}")
    print(f"- Một số nút trung tâm: {list(center)[:5]}")
    print(f"- Một số nút ngoại vi: {list(periphery)[:5]}")
else:
    print("\nĐồ thị không liên thông, không thể tính đường kính và bán kính.")
    print("Sử dụng thành phần liên thông lớn nhất để tính toán.")

print("\n" + "="*60)
````

## 2.4 Hệ số phân cụm (Clustering Coefficient)

````python
# Tính hệ số phân cụm
print("="*60)
print("HỆ SỐ PHÂN CỤM")
print("="*60)

# Hệ số phân cụm toàn cục (transitivity)
global_clustering = nx.transitivity(G)
print(f"\n- Hệ số phân cụm toàn cục (Transitivity): {global_clustering:.4f}")

# Hệ số phân cụm trung bình (average clustering coefficient)
avg_clustering = nx.average_clustering(G)
print(f"- Hệ số phân cụm trung bình: {avg_clustering:.4f}")

# Hệ số phân cụm cho từng nút
clustering_coeffs = nx.clustering(G)
clustering_values = list(clustering_coeffs.values())

print(f"\n- Hệ số phân cụm min: {min(clustering_values):.4f}")
print(f"- Hệ số phân cụm max: {max(clustering_values):.4f}")
print(f"- Hệ số phân cụm median: {np.median(clustering_values):.4f}")

print("\n" + "="*60)

# Vẽ histogram phân phối hệ số phân cụm
fig, ax = plt.subplots(figsize=(10, 6))
ax.hist(clustering_values, bins=30, edgecolor='black', alpha=0.7)
ax.axvline(avg_clustering, color='red', linestyle='--', linewidth=2, label=f'Trung bình = {avg_clustering:.4f}')
ax.set_xlabel('Hệ số phân cụm', fontsize=12)
ax.set_ylabel('Số lượng nút', fontsize=12)
ax.set_title('Phân phối Hệ số phân cụm', fontsize=14, fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"\nNhận xét: Giá trị trung bình là {avg_clustering:.4f}, cho thấy ", end="")
if avg_clustering > 0.5:
    print("mạng có xu hướng phân cụm cao (nút có xu hướng tạo thành các nhóm chặt chẽ)")
elif avg_clustering > 0.2:
    print("mạng có xu hướng phân cụm trung bình")
else:
    print("mạng có xu hướng phân cụm thấp (cấu trúc phân tán - đặc trưng của mạng scale-free)")
````

## 2.5 Độ dài đường đi trung bình

````python
# Tính độ dài đường đi trung bình
print("="*60)
print("ĐỘ DÀI ĐƯỜNG ĐI TRUNG BÌNH")
print("="*60)

if nx.is_connected(G_lcc):
    avg_path_length = nx.average_shortest_path_length(G_lcc)
    print(f"\n- Độ dài đường đi trung bình: {avg_path_length:.4f}")
    
    # Tính tất cả độ dài đường đi ngắn nhất
    all_shortest_paths = dict(nx.all_pairs_shortest_path_length(G_lcc))
    
    # Tạo danh sách các độ dài
    path_lengths = []
    for source in all_shortest_paths:
        for target, length in all_shortest_paths[source].items():
            if source != target:
                path_lengths.append(length)
    
    print(f"- Độ dài đường đi ngắn nhất min: {min(path_lengths)}")
    print(f"- Độ dài đường đi ngắn nhất max: {max(path_lengths)}")
    print(f"- Độ dài đường đi ngắn nhất median: {np.median(path_lengths):.2f}")
    
    # Vẽ histogram
    fig, ax = plt.subplots(figsize=(10, 6))
    counts, bins, patches = ax.hist(path_lengths, bins=range(1, max(path_lengths)+2), 
                                      edgecolor='black', alpha=0.7, align='left')
    ax.axvline(avg_path_length, color='red', linestyle='--', linewidth=2, 
               label=f'Trung bình = {avg_path_length:.4f}')
    ax.set_xlabel('Độ dài đường đi ngắn nhất', fontsize=12)
    ax.set_ylabel('Tần suất', fontsize=12)
    ax.set_title('Phân phối Độ dài đường đi ngắn nhất', fontsize=14, fontweight='bold')
    ax.legend()
    ax.grid(True, alpha=0.3, axis='y')
    plt.tight_layout()
    plt.show()
    
    print(f"\nNhận xét: Độ dài đường đi trung bình là {avg_path_length:.4f}, ", end="")
    if avg_path_length < 3:
        print("cho thấy mạng có đặc tính 'small-world' rất mạnh")
    elif avg_path_length < 5:
        print("cho thấy mạng có đặc tính 'small-world' (đặc trưng mạng scale-free)")
    else:
        print("cho thấy mạng có cấu trúc phân tán")
else:
    print("\nĐồ thị không liên thông. Tính cho thành phần liên thông lớn nhất:")
    avg_path_length = nx.average_shortest_path_length(G_lcc)
    print(f"- Độ dài đường đi trung bình (LCC): {avg_path_length:.4f}")

print("\n" + "="*60)
````

## 2.6 Phân phối bậc và Mô hình hồi quy

````python
# Phân tích phân phối bậc
print("="*60)
print("PHÂN PHỐI BẬC")
print("="*60)

# Lấy danh sách bậc của các nút
degrees = [G.degree(n) for n in G.nodes()]
degree_counts = Counter(degrees)

print(f"\n- Bậc trung bình: {np.mean(degrees):.2f}")
print(f"- Bậc median: {np.median(degrees):.2f}")
print(f"- Bậc min: {min(degrees)}")
print(f"- Bậc max: {max(degrees)}")
print(f"- Độ lệch chuẩn bậc: {np.std(degrees):.2f}")

# Vẽ phân phối bậc
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Histogram bậc
axes[0].hist(degrees, bins=30, edgecolor='black', alpha=0.7)
axes[0].axvline(np.mean(degrees), color='red', linestyle='--', linewidth=2, 
                label=f'Trung bình = {np.mean(degrees):.2f}')
axes[0].set_xlabel('Bậc', fontsize=12)
axes[0].set_ylabel('Số lượng nút', fontsize=12)
axes[0].set_title('Phân phối Bậc (Linear scale)', fontsize=14, fontweight='bold')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Log-log plot để kiểm tra power-law
degree_values = sorted(degree_counts.keys())
degree_freq = [degree_counts[d] for d in degree_values]

axes[1].scatter(degree_values, degree_freq, alpha=0.6, s=50)
axes[1].set_xlabel('Bậc (log scale)', fontsize=12)
axes[1].set_ylabel('Tần suất (log scale)', fontsize=12)
axes[1].set_title('Phân phối Bậc (Log-Log scale)\n→ Đường thẳng = Power-law (Scale-free)', fontsize=14, fontweight='bold')
axes[1].set_xscale('log')
axes[1].set_yscale('log')
axes[1].grid(True, alpha=0.3, which='both')

plt.tight_layout()
plt.show()

print("\n" + "="*60)
````

### Hồi quy Power-law

````python
# Hồi quy Power-law: P(k) ~ k^(-gamma)
# Sử dụng log transformation: log(P(k)) = -gamma * log(k) + c

# Lọc bỏ các giá trị 0
degree_values_nonzero = [d for d in degree_values if d > 0 and degree_counts[d] > 0]
degree_freq_nonzero = [degree_counts[d] for d in degree_values_nonzero]

if len(degree_values_nonzero) > 1:
    # Chuyển sang log scale
    log_degrees = np.log10(degree_values_nonzero)
    log_freq = np.log10(degree_freq_nonzero)
    
    # Hồi quy tuyến tính
    slope, intercept, r_value, p_value, std_err = stats.linregress(log_degrees, log_freq)
    
    print("="*60)
    print("HỒI QUY POWER-LAW")
    print("="*60)
    print(f"\nMô hình: P(k) = C * k^γ")
    print(f"\n- Hệ số gamma (γ): {-slope:.4f}")
    print(f"- Hệ số tương quan R²: {r_value**2:.4f}")
    print(f"- P-value: {p_value:.6f}")
    
    # Vẽ đồ thị hồi quy
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.scatter(degree_values_nonzero, degree_freq_nonzero, alpha=0.6, s=50, label='Dữ liệu thực')
    
    # Đường hồi quy
    x_fit = np.logspace(np.log10(min(degree_values_nonzero)), 
                        np.log10(max(degree_values_nonzero)), 100)
    y_fit = 10**(slope * np.log10(x_fit) + intercept)
    ax.plot(x_fit, y_fit, 'r--', linewidth=2, 
            label=f'Power-law fit: γ={-slope:.2f}, R²={r_value**2:.3f}')
    
    ax.set_xlabel('Bậc (log scale)', fontsize=12)
    ax.set_ylabel('Tần suất (log scale)', fontsize=12)
    ax.set_title('Hồi quy Power-law cho Phân phối Bậc', fontsize=14, fontweight='bold')
    ax.set_xscale('log')
    ax.set_yscale('log')
    ax.legend()
    ax.grid(True, alpha=0.3, which='both')
    plt.tight_layout()
    plt.show()
    
    print(f"\nNhận xét:")
    if r_value**2 > 0.8:
        print(f"- Mạng tuân theo phân phối power-law rất tốt (R² = {r_value**2:.3f})")
        print(f"- Đây là mạng SCALE-FREE với gamma = {-slope:.2f}")
        print(f"- Đặc trưng: Có vài hub nodes (bậc rất cao) và nhiều nút bậc thấp")
    elif r_value**2 > 0.6:
        print(f"- Mạng có xu hướng power-law khá tốt (R² = {r_value**2:.3f})")
    else:
        print(f"- Mạng không tuân theo phân phối power-law rõ ràng (R² = {r_value**2:.3f})")
    
    print("\n" + "="*60)
else:
    print("Không đủ dữ liệu để thực hiện hồi quy power-law")
````

## 2.7 Bố cục mạng (Network Layout)

**Lưu ý:** Để visualization tốt hơn với màu sắc và kích thước node, bạn nên sử dụng **Gephi**. 
Code dưới đây xuất file để import vào Gephi.

````python
# Visualization cơ bản trong Python
fig, axes = plt.subplots(2, 2, figsize=(16, 16))

# Layout 1: Spring layout (Force-directed)
pos1 = nx.spring_layout(G, k=0.5, iterations=50, seed=42)
node_sizes1 = [G.degree(n) * 20 for n in G.nodes()]
axes[0, 0].set_title('Spring Layout (Force-directed)', fontsize=14, fontweight='bold')
nx.draw_networkx(G, pos=pos1, ax=axes[0, 0], node_size=node_sizes1, 
                 node_color='lightblue', edge_color='gray', 
                 with_labels=False, alpha=0.7, width=0.5)
axes[0, 0].axis('off')

# Layout 2: Circular layout
pos2 = nx.circular_layout(G)
node_sizes2 = [G.degree(n) * 20 for n in G.nodes()]
axes[0, 1].set_title('Circular Layout', fontsize=14, fontweight='bold')
nx.draw_networkx(G, pos=pos2, ax=axes[0, 1], node_size=node_sizes2, 
                 node_color='lightgreen', edge_color='gray', 
                 with_labels=False, alpha=0.7, width=0.5)
axes[0, 1].axis('off')

# Layout 3: Kamada-Kawai layout
pos3 = nx.kamada_kawai_layout(G)
node_sizes3 = [G.degree(n) * 20 for n in G.nodes()]
axes[1, 0].set_title('Kamada-Kawai Layout', fontsize=14, fontweight='bold')
nx.draw_networkx(G, pos=pos3, ax=axes[1, 0], node_size=node_sizes3, 
                 node_color='lightcoral', edge_color='gray', 
                 with_labels=False, alpha=0.7, width=0.5)
axes[1, 0].axis('off')

# Layout 4: Shell layout
pos4 = nx.shell_layout(G)
node_sizes4 = [G.degree(n) * 20 for n in G.nodes()]
axes[1, 1].set_title('Shell Layout', fontsize=14, fontweight='bold')
nx.draw_networkx(G, pos=pos4, ax=axes[1, 1], node_size=node_sizes4, 
                 node_color='plum', edge_color='gray', 
                 with_labels=False, alpha=0.7, width=0.5)
axes[1, 1].axis('off')

plt.tight_layout()
plt.show()

print("Kích thước node tỷ lệ với bậc của nút")
````
