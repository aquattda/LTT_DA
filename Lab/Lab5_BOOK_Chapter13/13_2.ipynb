## 2. Ghép Dữ liệu MovieLens

```python
data = pd.merge(pd.merge(ratings, users), movies)
data.info()
```

**Nhận xét:**
- Dữ liệu sau khi gộp 3 bảng, chứa 1.000.209 hàng và 10 cột, cho phép truy cập tất cả thông tin về user, rating, và movie trong cùng một hàng.
- Việc gộp dữ liệu làm dễ dàng hơn việc phân tích giữa các cột ở các bảng khác nhau một cách dễ dàng.

## 3. Phân tích khám phá

### 3.1 Tính điểm trung bình 

```python
mean_ratings = data.pivot_table("rating", index="title",
                                columns="gender", aggfunc="mean")
mean_ratings.head(5)
```

**Nhận xét:**
- Trước hết có thể thấy rằng thứ tự xếp hạng trung bình (rating) cho mỗi tiêu đề phim (title) được nhóm theo giới tính (gender). 
- Dựa vào kết quả cho thấy nữ giới chiếm tỷ lệ xem cao hơn so với nam giới về các bộ phim.

### 3.2. Lọc phim phổ biến

```python
ratings_by_title = data.groupby("title").size()
ratings_by_title.head()
active_titles = ratings_by_title.index[ratings_by_title >= 250]
active_titles
```

```python
mean_ratings = mean_ratings.loc[active_titles]
mean_ratings
```

**Nhận xét:**
- Thực hiện nhóm theo title và chỉ lấy số lượng phim ít nhất là 250 lượt xếp hạng.

Ví dụ, một bộ phim có 3 lượt xếp hạng (rating) và nhận được điểm trung bình là 5.0 sẽ không đáng tin cậy bằng một bộ phim có 250 lượt xếp hạng và điểm trung bình là 4.5. Việc đặt ngưỡng tối thiểu 250 đảm bảo rằng mỗi bộ phim được đưa vào phân tích tiếp theo (ví dụ: so sánh xếp hạng theo giới tính, tìm độ bất đồng) đều có một cơ sở dữ liệu đủ lớn.

**1) Độ lệch chuẩn (std) của các xếp hạng cho mỗi bộ phim**

```python
mean_ratings["diff"] = mean_ratings["M"] - mean_ratings["F"]
sorted_by_diff = mean_ratings.sort_values("diff")
sorted_by_diff.head()
sorted_by_diff[::-1].head()
rating_std_by_title = data.groupby("title")["rating"].std()
rating_std_by_title = rating_std_by_title.loc[active_titles]
rating_std_by_title.head()
```

**Nhận xét:**
- Những tựa phim trên có mức phân cực ý kiến cao trong toàn bộ người xem (không riêng theo giới), phản ánh khác biệt về khẩu vị (ví dụ: hài tuổi teen, remake, phong cách cũ/đen-trắng,…).

- 'burbs, The (1989) đứng đầu → mức độ “hoặc thích hoặc ghét” rõ rệt nhất trong danh sách đang xét.

**2) “Liệu các thể loại phim cực đoan (kinh dị, tâm lý, chính trị) có xu hướng có độ lệch chuẩn rating cao hơn so với phim tình cảm, hài hay không?”.**

```python
rating_std_by_title.sort_values(ascending=False)[:10]
```

**Nhận xét:**
- Có cơ sở cho thấy các thể loại “phân cực” (đặc biệt Horror/Kinh dị, Psychological/Arthouse như Eyes Wide Shut, Documentary, Cult/Experimental) có độ lệch chuẩn rating cao hơn so với các thể loại “an toàn” như Romance hay Comedy. Danh sách top phim lệch chuẩn cao bạn in ra gồm nhiều tác phẩm kinh dị/thử nghiệm/cult (Blair Witch Project, Rocky Horror, Fear and Loathing…)—đây là các phim thường “hoặc rất thích hoặc rất ghét”, nên phân tán điểm lớn.

Điểm chung nổi bật:

>Các thể loại/thiết kế này thường tạo kỳ vọng khác nhau theo giới về nội dung, nhịp điệu và phong cách kể chuyện, do đó gây chênh lệch rõ rệt về mức yêu thích.

**2) “Người dùng trẻ tuổi (dưới 25) có xu hướng đánh giá cực đoan (1 hoặc 5) hơn so với nhóm lớn tuổi không?”**

```python
movies["genres"].head()
movies["genres"].head().str.split("|")
movies["genre"] = movies.pop("genres").str.split("|")
movies.head()
```

```python
movies_exploded = movies.explode("genre")
movies_exploded[:10]
```

**Nhận xét:**
- Sử dụng phương thức explode để biến đổi DataFrame: mỗi phần tử trong danh sách thể loại sẽ tạo ra một hàng riêng biệt với cùng thông tin phim.

```python
ratings_with_genre = (
    movies_exploded.merge(ratings, on="movie_id")
                   .merge(users[["user_id","age"]], on="user_id")
)
age_to_bin = {
    1:  "1-18",     
    18: "18-25",   
    25: "25-35",  
    35: "35-45",  
    45: "45-56",   
    50: "45-56",   
    56: "45-56",
}
ratings_with_genre["age_bin"] = ratings_with_genre["age"].map(age_to_bin)

keep_bins = ["1-18", "18-25", "25-35", "35-45", "45-56"]
ratings_with_genre = ratings_with_genre[ratings_with_genre["age_bin"].isin(keep_bins)]
# mean
genre_ratings = (
    ratings_with_genre.groupby(["genre", "age_bin"])["rating"].mean().unstack("age_bin").reindex(columns=keep_bins)
)

# Xem 10 dòng đầu
genre_ratings.head(10)
```

**Nhận xét:**
- Hầu hết các thể loại (Action, Adventure, Animation, Children’s, Comedy, Crime, Documentary, Drama, Fantasy, Film-Noir) đều có xu hướng điểm trung bình tăng theo tuổi: các cột 1–18 và 18–25 thấp hơn rõ so với 35–45 và 45–56 (mức chênh ~0.1–0.3 điểm tùy thể loại).

>Kết luận trực tiếp từ bảng này: người dùng trẻ (<25) không cho thấy xu hướng “cực đoan” hay “hào phóng”; họ đơn giản là khắt khe hơn (điểm trung bình thấp hơn) so với nhóm lớn tuổi.
